### ⭐ Exercise 2 — "Friend Finder" 👯
Create vectors for 4 of your real friends: `[cricket, movies, gaming, foodie]` rated 0–10. Compute
cosine similarity between every pair. Does the math's "most similar pair" match your intuition?

Answer: I created vectors for four friends based on their interests in cricket,
       movies, gaming, and food, with values ranging from 0 to 10.

      Then I calculated the cosine similarity between every pair of friends. The highest similarity was between Rahul and Amit (~0.9995), followed by Neha and Riya (~0.9873).

      This matches my intuition because Rahul and Amit have very similar interests in cricket and gaming, while Neha and Riya share similar preferences for movies and food.

      

```python
import numpy as np

Rahul = np.array([8, 6, 7, 5])
Amit  = np.array([7, 5, 6, 4])
Neha  = np.array([2, 9, 3, 8])
Riya  = np.array([3, 8, 2, 9])

def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

ra = cosine_sim(Rahul, Amit)
nr = cosine_sim(Neha, Riya)

print(f"Rahul-Amit → {ra:.4f} (highest) ✅")
print(f"Neha-Riya → {nr:.4f} (very close)")

print("\n👉 Explanation:")
print("Rahul & Amit have almost the same interests")
print("Neha & Riya are also very similar")
```

### ⭐ Exercise 3 — "My Restaurant Brain" ⚖️
Write down YOUR true weights for `[taste, price, distance, ambience]` (must sum to 1). Score 3 real
restaurants near you and compute your dot-product score for each. Did the ranking match where you
actually like to go? If not — adjust the weights until it does. (You just
*trained* yourself!)

Answer: The restaurant score are calculated using the dot product of weights
        and features.among all option **Chocolate Room** has the highest score.

        Best Choice:
            Chocolate Room with a score of 8.0

        Conclusion:
              Chocolate Room is the most preferred restaurant based on the given weights.

```python
import numpy as np

# Step 1: Weights (must sum to 1)
# [taste, price, distance, ambience]
weights = np.array([0.4, 0.2, 0.1, 0.3])

# Step 2: Real restaurant scores (out of 10)
# Barbeque Nation
r1 = np.array([9, 6, 5, 8])

# Chocolate Room
r2 = np.array([8, 7, 7, 9])

# Sankalp
r3 = np.array([7, 8, 6, 6])

# Step 3: Dot product scores
s1 = np.dot(weights, r1)
s2 = np.dot(weights, r2)
s3 = np.dot(weights, r3)

# Step 4: Print scores
print("Restaurant Scores:")
print("Barbeque Nation:", round(s1, 2))
print("Chocolate Room:", round(s2, 2))
print("Sankalp:", round(s3, 2))

# Step 5: Ranking
scores = {
    "Barbeque Nation": s1,
    "Chocolate Room": s2,
    "Sankalp": s3
}

sorted_scores = sorted(scores.items(), key=lambda x: x[1], reverse=True)

print("\nRanking:")
for i, (name, score) in enumerate(sorted_scores, 1):
    print(f"{i}. {name} -> {round(score, 2)}")

# Step 7: Print highest scoring restaurant separately
top_restaurant, top_score = sorted_scores[0]

print("\nBest Choice:")
print(f"{top_restaurant} is the best option with a score of {round(top_score, 2)}")

# Step 6: Final Answer
print("\nConclusion:")
print("The highest score indicates my preferred restaurant based on my weights.")
```

### ⭐⭐ Exercise 4 — "Chai Quality Inspector" ☕
Simulate 100 days of chai with `rng.normal(100, 5, 100)`. Now append 3 'bad days': 70, 130, 145.
Compute z-scores for all 103 days and automatically print only the anomalies (|z| > 2.5).
Congratulations — you've built a fraud-detector skeleton.

Answer:  
      We generated 100 chai values using a Normal Distribution around 100 ml and added abnormal values (70, 130, 145). Using Z-score, we measured how far each value is from the mean.

      Values with |z| > 2.5 were flagged as anomalies.

       Result: 70, 130, and 145 are detected as unusual chai days.

```python
import numpy as np

# Step 1: Random generator
rng = np.random.default_rng(42)

# Step 2: Simulate 100 normal chai days (mean=100, std=5)
chai = rng.normal(100, 5, 100)

# Step 3: Add 3 abnormal days
chai = np.append(chai, [70, 130, 145])

# Step 4: Calculate mean and standard deviation
mean = np.mean(chai)
std = np.std(chai)

# Step 5: Compute z-scores
z_score = (chai - mean) / std

# Step 6: Print anomalies (|z| > 2.5)
print("Anomalies (|z| > 2.5):\n")
#print("Anomalies (|z| > 2.5):\n")

for i in range(len(chai)):
    if abs(z_score[i]) > 2.5:
       print(f"Day {i+1}: {chai[i]:.2f} ml (z = {z_score[i]:.2f})")

# Step 7: Conclusion
print("\nConclusion: Values with |z| > 2.5 are detected as unusual chai days.")
```

### ⭐⭐ Exercise 5 — "Mood Detector Neuron" 🤖
Build a neuron that detects if a message is ANGRY using features
`[count of '!!', count of CAPS words, contains 'sorry'?]`. Choose the weights yourself (should
'sorry' get a positive or negative weight?). Test it on 3 made-up messages with `sigmoid`.

Answer:
        Conclusion:
              Value close to 1 → Angry
              Value close to 0 → Not Angry
              "sorry" has negative weight because it reduces anger
              Message 1 is angry
              Message 2 is normal
              Message 3 is not angry (apology)


```python
import numpy as np

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Weights: [!! count, CAPS words, contains "sorry"]
weights = np.array([1.5, 1.0, -2.0])   # sorry is negative (reduces anger)
bias = -1

# Neuron function
def mood_neuron(x):
    z = np.dot(weights, x) + bias
    return sigmoid(z)

# --------------------------------
# Test Messages
# --------------------------------

# 1. Angry message
# "WHAT IS THIS!! THIS IS BAD!!"
m1 = np.array([2, 3, 0])

# 2. Normal message
# "Hello, how are you?"
m2 = np.array([0, 0, 0])

# 3. Apology message
# "SORRY, my mistake"
m3 = np.array([0, 1, 1])

# --------------------------------
# Output
# --------------------------------

print("Message 1 (Angry):", mood_neuron(m1))
print("Message 2 (Normal):", mood_neuron(m2))
print("Message 3 (Apology):", mood_neuron(m3))
```

### ⭐⭐ Exercise 6 — "Temperature Knob" 🌡️
Take scores `[5.1, 2.3, 0.8]` and compute `softmax(scores / T)` for T = 0.5, 1, 2, 5. Plot the bars
for each T. What happens to the model's "confidence" as T grows? (You've just discovered the
*temperature* setting in ChatGPT's API.)

Answer:
     Conclusion:
        Low temperature (T = 0.5) → sharp distribution → high confidence
        High temperature (T = 5) → flat distribution → low confidence

        As temperature increases, probabilities become more equal, so the model’s confidence decreases.


```python
import numpy as np
import matplotlib.pyplot as plt

# Softmax function
def softmax(x):
    e = np.exp(x - np.max(x))
    return e / np.sum(e)

scores = np.array([5.1, 2.3, 0.8])
temperatures = [0.5, 1, 2, 5]

# Plot
for T in temperatures:
    probs = softmax(scores / T)
    plt.figure()
    plt.bar(['A','B','C'], probs)
    plt.title(f"Softmax with T = {T}")
    print(f"T={T} →", probs)

plt.show()
```

### ⭐⭐⭐ Exercise 7 — "Grow the Search Engine" 🔎
Extend the semantic-search demo: add 4 more documents and invent a 4th embedding dimension
(e.g. 'travel-ness'). Write a function `search(query_vec, top_k)` returning the top-k documents.
Test 3 different queries. Bonus: make it refuse to answer (return nothing) if the best similarity
is below 0.5 — you've just added a "relevance threshold," a real RAG technique.

Answer:  
      Added 4th dimension (travel) to embeddings
      search(query_vec, top_k) returns top-k similar documents
      If similarity < 0.5 → returns empty list (no answer)
      This is called a relevance threshold, used in real-world search/RAG systems

```python
import numpy as np

# Cosine similarity function
def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Documents with 4D embeddings: [food, tech, sports, travel]
docs = {
    "Pizza recipe":        np.array([0.9, 0.1, 0.0, 0.1]),
    "AI tutorial":         np.array([0.1, 0.9, 0.1, 0.0]),
    "Cricket match":       np.array([0.0, 0.1, 0.9, 0.1]),
    "Travel guide Paris":  np.array([0.1, 0.0, 0.1, 0.9]),
    "Burger review":       np.array([0.8, 0.1, 0.0, 0.1]),
    "Laptop specs":        np.array([0.0, 0.9, 0.1, 0.0]),
    "Football news":       np.array([0.0, 0.1, 0.8, 0.1]),
    "Beach vacation":      np.array([0.1, 0.0, 0.1, 0.9])
}

# Search function with top-k and threshold
def search(query_vec, top_k=2, threshold=0.5):
    similarities = []

    for name, vec in docs.items():
        sim = cosine_sim(query_vec, vec)
        similarities.append((name, sim))

    # Sort by highest similarity
    similarities.sort(key=lambda x: x[1], reverse=True)

    # Apply relevance threshold
    if similarities[0][1] < threshold:
        return []

    return similarities[:top_k]

# -----------------------
# Test Queries
# -----------------------

# Query 1: Food
q1 = np.array([1, 0, 0, 0])

# Query 2: Tech
q2 = np.array([0, 1, 0, 0])

# Query 3: Weak/irrelevant
q3 = np.array([0, 0, 0, 0.1])

# Results
print("Query 1 (Food):", search(q1))
print("Query 2 (Tech):", search(q2))
print("Query 3 (Low relevance):", search(q3))
```

### ⭐⭐⭐ Exercise 8 — "Attention Detective" 🕵️
Re-run the attention demo, but make the focus word "cat" instead of "sat". Which word does "cat"
attend to most? Then *edit the vector* of "the" to make it more similar to "cat" and watch the
attention weights shift. Write 2 sentences on what this tells you about how GPT decides relevance.


Answer:
      Attention Detective:

      When the focus word is "cat", it attends most to "cat" and "mat" because their vectors are most similar.

      After modifying the vector of "the" to be closer to "cat", its attention weight increases, so "the" gets more focus.

      Conclusion:

      Attention depends on vector similarity between words.
      If similarity increases, attention weight also increases, showing how GPT decides relevance.  

```python
import numpy as np

# Softmax function
def softmax(x):
    e = np.exp(x - np.max(x))
    return e / np.sum(e)

# Words
words = ["the", "cat", "sat", "on", "mat"]

# Embeddings
embeddings = {
    "the": np.array([0.1, 0.1]),
    "cat": np.array([0.9, 0.8]),
    "sat": np.array([0.8, 0.7]),
    "on":  np.array([0.2, 0.1]),
    "mat": np.array([0.7, 0.6])
}

# Focus word = "cat"
query = embeddings["cat"]

# Attention scores
scores = [np.dot(query, embeddings[w]) for w in words]

# Attention weights
attention = softmax(np.array(scores))

print("Words:", words)
print("Attention:", attention)

# Modify "the" vector to be similar to "cat"
embeddings["the"] = np.array([0.8, 0.7])

# Recompute attention
scores = [np.dot(query, embeddings[w]) for w in words]
attention = softmax(np.array(scores))

print("New Attention:", attention)
```
